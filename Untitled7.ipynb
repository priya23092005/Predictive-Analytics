```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# --- 1. Create a Sample Dataset (Replace with your actual data loading) ---
# In a real scenario, you would load your data from a CSV, Excel, database, etc.
# Example: df = pd.read_csv('your_admission_data.csv')

data = {
    'GPA': np.random.uniform(2.5, 4.0, 200).round(2),
    'SAT_Score': np.random.randint(1000, 1600, 200),
    'Extracurriculars': np.random.randint(0, 5, 200),
    'Essays_Score': np.random.uniform(60, 100, 200).round(1),
    'Recommendation_Letters': np.random.choice(['Poor', 'Average', 'Good', 'Excellent'], 200),
    'Previous_School_Rank': np.random.randint(1, 10, 200), # 1 being top rank
    'Admission_Status': np.random.choice(['Admitted', 'Rejected'], 200, p=[0.4, 0.6])
}
df = pd.DataFrame(data)

# Introduce some correlation for demonstration (e.g., higher GPA/SAT -> more likely admitted)
df.loc[df['GPA'] > 3.5, 'Admission_Status'] = np.random.choice(['Admitted', 'Rejected'], df[df['GPA'] > 3.5].shape[0], p=[0.7, 0.3])
df.loc[df['SAT_Score'] > 1400, 'Admission_Status'] = np.random.choice(['Admitted', 'Rejected'], df[df['SAT_Score'] > 1400].shape[0], p=[0.7, 0.3])
df.loc[df['Admission_Status'] == 'Admitted', 'GPA'] = df.loc[df['Admission_Status'] == 'Admitted', 'GPA'] + np.random.uniform(0, 0.5, df[df['Admission_Status'] == 'Admitted'].shape[0])
df.loc[df['Admission_Status'] == 'Admitted', 'SAT_Score'] = df.loc[df['Admission_Status'] == 'Admitted', 'SAT_Score'] + np.random.randint(0, 100, df[df['Admission_Status'] == 'Admitted'].shape[0])
df['GPA'] = df['GPA'].clip(2.5, 4.0).round(2) # Clip to keep GPA within bounds
df['SAT_Score'] = df['SAT_Score'].clip(1000, 1600).astype(int) # Clip to keep SAT within bounds

print("--- Sample Data Head ---")
print(df.head())
print("\n--- Data Info ---")
df.info()
print("\n--- Basic Statistics for Numerical Features ---")
print(df.describe())

# --- 2. Descriptive Analysis ---

print("\n--- Admission Status Distribution ---")
print(df['Admission_Status'].value_counts())
print(df['Admission_Status'].value_counts(normalize=True) * 100) # Percentage

plt.figure(figsize=(6, 4))
sns.countplot(x='Admission_Status', data=df, palette='viridis')
plt.title('Distribution of Admission Status')
plt.xlabel('Admission Status')
plt.ylabel('Number of Students')
plt.show()

# --- 3. Analyzing Numerical Features by Admission Status ---

print("\n--- GPA Statistics by Admission Status ---")
print(df.groupby('Admission_Status')['GPA'].describe())

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='GPA', hue='Admission_Status', kde=True, palette='coolwarm')
plt.title('GPA Distribution by Admission Status')
plt.xlabel('GPA')
plt.ylabel('Count')
plt.show()

print("\n--- SAT Score Statistics by Admission Status ---")
print(df.groupby('Admission_Status')['SAT_Score'].describe())

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='SAT_Score', hue='Admission_Status', kde=True, palette='coolwarm')
plt.title('SAT Score Distribution by Admission Status')
plt.xlabel('SAT Score')
plt.ylabel('Count')
plt.show()

print("\n--- Essays Score Statistics by Admission Status ---")
print(df.groupby('Admission_Status')['Essays_Score'].describe())

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Essays_Score', hue='Admission_Status', kde=True, palette='coolwarm')
plt.title('Essays Score Distribution by Admission Status')
plt.xlabel('Essays Score')
plt.ylabel('Count')
plt.show()


# --- 4. Analyzing Categorical/Ordinal Features by Admission Status ---

print("\n--- Extracurriculars by Admission Status ---")
# Count plot for Extracurriculars
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Extracurriculars', hue='Admission_Status', palette='pastel')
plt.title('Extracurricular Activities by Admission Status')
plt.xlabel('Number of Extracurriculars')
plt.ylabel('Number of Students')
plt.show()

# Crosstab for Extracurriculars (showing counts and percentages)
print(pd.crosstab(df['Extracurriculars'], df['Admission_Status']))
print(pd.crosstab(df['Extracurriculars'], df['Admission_Status'], normalize='index').round(2) * 100) # Row percentages

print("\n--- Recommendation Letters by Admission Status ---")
# Order for plotting
rec_order = ['Poor', 'Average', 'Good', 'Excellent']
df['Recommendation_Letters'] = pd.Categorical(df['Recommendation_Letters'], categories=rec_order, ordered=True)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Recommendation_Letters', hue='Admission_Status', palette='muted')
plt.title('Recommendation Letters Quality by Admission Status')
plt.xlabel('Recommendation Letter Quality')
plt.ylabel('Number of Students')
plt.show()

print(pd.crosstab(df['Recommendation_Letters'], df['Admission_Status']))
print(pd.crosstab(df['Recommendation_Letters'], df['Admission_Status'], normalize='index').round(2) * 100)

print("\n--- Previous School Rank by Admission Status ---")
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Previous_School_Rank', hue='Admission_Status', palette='deep')
plt.title('Previous School Rank by Admission Status')
plt.xlabel('Previous School Rank (1 = Best)')
plt.ylabel('Number of Students')
plt.show()

print(pd.crosstab(df['Previous_School_Rank'], df['Admission_Status']))
print(pd.crosstab(df['Previous_School_Rank'], df['Admission_Status'], normalize='index').round(2) * 100)

# --- 5. Correlation Analysis (for numerical features) ---
print("\n--- Correlation Matrix of Numerical Features ---")
numerical_cols = ['GPA', 'SAT_Score', 'Extracurriculars', 'Essays_Score', 'Previous_School_Rank']
correlation_matrix = df[numerical_cols].corr()
print(correlation_matrix)

plt.figure(figsize=(8, 7))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show()

# --- 6. Pairplot for selected numerical features (can be computationally intensive for many features) ---
print("\n--- Pairplot of Numerical Features by Admission Status (might take a moment) ---")
# Select a subset of numerical features if you have many, for better readability
selected_numerical_for_pairplot = ['GPA', 'SAT_Score', 'Essays_Score', 'Admission_Status']
sns.pairplot(df[selected_numerical_for_pairplot], hue='Admission_Status', palette='viridis')
plt.suptitle('Pairplot of Numerical Features by Admission Status', y=1.02)
plt.show()
```
