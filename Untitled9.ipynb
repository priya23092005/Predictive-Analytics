```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load sample loan dataset (You can replace with your own CSV file)
# Dataset should have columns like 'Loan_Status', 'ApplicantIncome', 'Credit_History', etc.
df = pd.read_csv('loan_data.csv')

# 1. Overview of the dataset
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:\n", df.head())
print("\nSummary statistics:\n", df.describe(include='all'))

# 2. Count of approved vs rejected loans
print("\nLoan Status counts:\n", df['Loan_Status'].value_counts())

# Visualize loan approval distribution
sns.countplot(x='Loan_Status', data=df)
plt.title('Loan Approval Distribution')
plt.show()

# 3. Relationship between credit history and loan approval
print("\nLoan approval based on Credit History:")
print(pd.crosstab(df['Credit_History'], df['Loan_Status'], normalize='index'))

sns.countplot(x='Credit_History', hue='Loan_Status', data=df)
plt.title('Loan Status by Credit History')
plt.show()

# 4. Income distribution by loan status
sns.boxplot(x='Loan_Status', y='ApplicantIncome', data=df)
plt.title('Applicant Income by Loan Status')
plt.show()

# 5. Descriptive statistics for income by loan status
print("\nIncome statistics by Loan Status:")
print(df.groupby('Loan_Status')['ApplicantIncome'].describe())

# 6. Correlation heatmap for numeric variables
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
plt.figure(figsize=(8,6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Numeric Features')
plt.show()

# Summary insight example:
print("\nInsight example:")
print(" - Most approved loans have Credit_History = 1")
print(" - Higher ApplicantIncome tends to be linked with approved loans")
print(" - Correlation heatmap shows the relationships between numeric features and loan approval")

```
